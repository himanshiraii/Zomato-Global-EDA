```python
# eliminating columns which are not important for EDA
df1=df.drop(["Restaurant ID","Address","Locality Verbose","Longitude","Latitude","Rating color","Switch to order menu"],axis=1)
```

```python
# sample of new updated dataset
df1.sample(5)
```

```python
# dropping duplicate values if exists
df1.drop_duplicates(inplace=True)
```

```python
# revised length of df1
len(df1)
```

```python
# checking for null values in new dataset
for k in df1.columns:
    if df1[k].isnull().sum()>0:
        print(f"{k} = {df1[k].isnull().sum()} \ndata type = {df1[k].dtype}")
```

```python
# filling null values 
df1["Cuisines"].fillna(df1["Cuisines"].mode()[0],inplace=True)
```

```python
# post removing duplicates & handling missing values dataset
df1.info()
```

```python
# nunique values in new dataset
for k in df1.columns:
    print(f"{k} = {df1[k].nunique()}")
```

# Note:

OUTLIER DETECTION AND REMOVAL

There will be restaurants in this dataset which will have high votes/Aggregate rating which indicates that such restaurants are popular and worthy. And using outlier detection or removal on such numerical restaurants will negatively impact the dataset because after applying outlier removal, missing values will be generated and if we fill those missing values with either the median or mean of the particular column, then it will convert a reputed restaurant into an average restaurant which can lead to misleading/wrong statistics of the restaurants
